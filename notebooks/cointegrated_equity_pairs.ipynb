The first step of my strategy is finding pairs of cointegrated equity stocks. To be good candidates for a mean-reversion model the underlying business model, market exposures and macroeconomic sensitivities must be as similar as possible. Although a mean-reversion should fundamentally be market-neutral, it is always best to pick stocks from a range of sectors to protect against any extreme market movements.

**Initial ideas**

I would like to primarily focus on US equities because this is typically where the biggest global giants are traded.

**Energy and materials**:

XOM and CVX are a pair of mega-cap oil companies. They are both fully integrated energy companies and are directly exposed Brent and WTI crude cycles.

COP and OXY are both heavily focussed on exploration and production, so their price actions are both highly dependant on domestic oil inventory data.

BHP and RIO are both global mining giants whose prices fluctuate based on iron and copper spot prices, and from Chinese industrial demand.

**Consumer retail**:

KO and PEP are the classic statistical arbitrage pair. The long-term equilibrium is dependant on global demand of the staple drink, regardless of short-term marketing or earnings divergences.

HD and LOW are a duopoly of home improvement retail. Their spreads are sensitive to US mortgage rates and housing supply.

WMT and TGT are broad-market consumer retailers. Any divergences will typically occur during earnings season and converge soon to the mean.

**Financials**:

JPM and BAC are both sensitive to the same yield curves and Federal Reserve rate decisions.

V and MA are both payment processing networks with very similar business models based on transaction volume.

WFC and USB are commercial banking giants heavily dependant on US domestic lending.

**Technology**:

GOOGL and GOOG is purely structural, where the only difference is voting rights.

CSCO and JNPR are legacy hardware competitors that often move in tandem with broader IT spending cycles.

MSFT and AAPL may have different business models, however their massive weight in the S&P 500 and QQQ often forces their ETF to move together.
